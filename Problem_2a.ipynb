# Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# Load and Display Coins Image

```python
# Read coins image as grayscale
coins = cv2.imread("coins.png", cv2.IMREAD_GRAYSCALE)

# Check image
print("Image shape:", coins.shape)
print("Minimum intensity:", coins.min())
print("Maximum intensity:", coins.max())

# Display image
plt.figure(figsize=(6, 5))
plt.imshow(coins, cmap="gray", vmin=0, vmax=255)
plt.title("Coins Image")
plt.axis("off")
plt.show()
```

# Compute and Plot 8-bit Histogram

```python
# Compute Histogram

# Frequency of each intensity level (0 to 255)
histogram = np.bincount(coins.ravel(), minlength=256)

# Intensity levels
intensity_levels = np.arange(256)

# Print frequency for each intensity level
# for intensity, frequency in enumerate(histogram):
#     print(f"Intensity {intensity}: {frequency} pixels")


# Plot Histogram

plt.figure(figsize=(10, 5))

plt.bar(intensity_levels, histogram, width=1)

plt.xlabel("Intensity Level")
plt.ylabel("Frequency (Number of Pixels)")
plt.title("Histogram of coins.png")
plt.xlim(0, 255)

plt.show()
```
