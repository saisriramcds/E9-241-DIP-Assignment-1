# Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# Load and Display Coins Image

```python
# Read coins image as grayscale
coins = cv2.imread("coins.png", cv2.IMREAD_GRAYSCALE)

# Check image
print("Image shape:", coins.shape)
print("Minimum intensity:", coins.min())
print("Maximum intensity:", coins.max())

# Display image
plt.figure(figsize=(6, 5))
plt.imshow(coins, cmap="gray", vmin=0, vmax=255)
plt.title("Coins Image")
plt.axis("off")
plt.show()
```

# Compute Histogram

```python
# Flatten the image into a 1D array
pixels = coins.ravel()

# Compute histogram for intensity levels 0 to 255
histogram = np.bincount(pixels, minlength=256)

# Intensity levels
intensity_levels = np.arange(256)

# Total number of pixels
N = len(pixels)

print("Total number of pixels:", N)
```

# Plot Histogram

```python
plt.figure(figsize=(10, 5))

plt.bar(intensity_levels, histogram, width=1)

plt.xlabel("Intensity Level")
plt.ylabel("Frequency")
plt.title("Histogram of coins.png")
plt.xlim(0, 255)

plt.show()
```

# Expectation and Variance Using Histogram

```python
# Intensity values
k = np.arange(256)

# Probability of each intensity level
probability = histogram / N

# Expectation (Mean) using histogram

mean_histogram = np.sum(k * probability)

# Variance using histogram

variance_histogram = np.sum((k - mean_histogram)**2 * probability)

# Display results
print("Using Histogram:")
print("-------------------------")
print("Expectation =", mean_histogram)
print("Variance    =", variance_histogram)
```

# Verify Using Actual Pixel Values

```python

# Mean using actual pixel values

mean_actual = np.mean(pixels)


# Variance using actual pixel values

variance_actual = np.var(pixels)

# Display results
print("Using Actual Pixel Values:")
print("-------------------------")
print("Expectation =", mean_actual)
print("Variance    =", variance_actual)
```

# Compare Histogram and Actual Values

```python
# Calculate differences
mean_difference = abs(mean_histogram - mean_actual)
variance_difference = abs(variance_histogram - variance_actual)

print("Comparison")
print("====================================")

print(f"Mean using histogram : {mean_histogram:.6f}")
print(f"Mean using pixels    : {mean_actual:.6f}")
print(f"Mean difference      : {mean_difference:.10f}")

print()

print(f"Variance using histogram : {variance_histogram:.6f}")
print(f"Variance using pixels    : {variance_actual:.6f}")
print(f"Variance difference      : {variance_difference:.10f}")
```
